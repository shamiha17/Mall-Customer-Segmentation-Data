```python
df['Gender']=df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
df['Gender']
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler= StandardScaler()
```

```python
df[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']]=scaler.fit_transform(df[['Age',	'Annual Income (k$)',	'Spending Score (1-100)']])
```

```python
df.columns
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python

enc = OneHotEncoder(drop='first', sparse_output=False)
gender_encoded = enc.fit_transform(df[['Gender']])
gender_df = pd.DataFrame(gender_encoded, columns=enc.get_feature_names_out(['Gender']))
```

```python
X = pd.concat([df[['Age', 'Annual Income (k$)']], gender_df], axis=1)
y = df['Spending Score (1-100)'] 
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,y)
```

```python
# Sklearn models
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor


# XGBoost

from xgboost import XGBRegressor
```

```python
models = {
    
     "Linear Regression": LinearRegression(),
    "Random Forest": RandomForestRegressor(random_state=42),
    
  "Xgb_model": XGBRegressor(random_state=42)


}
```

```python
for model_name in models.keys():
    models[model_name].fit(x_train,y_train)
    print(model_name)
    print( models[model_name].score(x_test,y_test))
    print(models[model_name].predict(x_test))
```

```python
preds=models[model_name].predict(x_test)
```
